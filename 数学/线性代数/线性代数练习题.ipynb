## 随机生成两个4×4矩阵，求下列指定的四个矩阵，并确定哪些矩阵是相等的

![](../../pic/Pasted%20image%2020260724114145.png)

```python
import numpy as np

A = np.random.randint(0, 10, size=(4, 4))
B = np.random.randint(0, 10, size=(4, 4))
```

```python
A1 = A @ B
A2 = B @ A
A3 = (A.T @ B.T).T
A4 = (B.T @ A.T).T
A1,A2, A3, A4
```

Numpy中使用@运算符或者np.dot 进行矩阵的乘法
矩阵间使用 * 运算符 表示按元素相乘
